# Entity linking: justificación de las decisiones

Cada sección compara solo las alternativas relevantes para una decisión del pipeline:

| # | Fase | Decisión | Se compara con |
|---|---|---|---|
| 1 | NER | GLiNER con tipos finos | etiquetas genéricas; LLM local |
| 2 | Retrieval | `wbsearchentities`, 20 candidatos | 5 candidatos |
| 3 | Desambiguación | criterio de incertidumbre + LLM solo en dudosas | primer candidato |
| 4 | End-to-end | reordenar por tipo fino + coherencia | orden de Wikidata |
| 5 | Texto libre | pipeline completo en inglés y español | – |

Datos: AIDA CoNLL-YAGO (`test`, noticias con todas las entidades anotadas y su QID) y textos anotados a mano (`manual_ner.jsonl`). `RUN_LLM` activa las celdas con LLM (~45 s por llamada en este equipo).

## 0. Configuración

```python
%load_ext autoreload
%autoreload 2

import json
import time
from html import escape

from IPython.display import HTML, display

from entity_linker.evaluation.datasets import (
    DATA_DIR,
    Document,
    GoldMention,
    load_aida,
    load_disambiguation_sample,
)
from entity_linker.evaluation.evaluate import (
    evaluate_disambiguation,
    evaluate_linking,
    evaluate_ner,
    evaluate_retrieval,
)
from entity_linker.services.entity_types import FINE_LABELS, GENERIC_LABELS
from entity_linker.services.pipeline import EntityLinkingPipeline

LIMIT = 10  # documentos de AIDA en las comparaciones
RUN_LLM = False  # True: ejecuta las celdas con LLM (Ollama, ~7 s por llamada en GPU)
```

```python
COLORS = {"PER": "#ffd6a5", "ORG": "#a0c4ff", "LOC": "#caffbf", "MISC": "#ffc6ff"}


def spans(entities):
    return {(e.start_char, e.end_char, e.label) for e in entities}


def gold_spans(doc):
    return {(m.start, m.end, m.label) for m in doc.mentions}


def highlight(text, spans_):
    """Texto con las entidades resaltadas, un color por etiqueta."""
    out, pos = [], 0
    for start, end, label in sorted(spans_, key=lambda s: s[:2]):
        out.append(escape(text[pos:start]))
        out.append(
            f'<mark style="background:{COLORS.get(label, "#ddd")};color:#000">'
            f"{escape(text[start:end])} <sub>{label}</sub></mark>"
        )
        pos = end
    out.append(escape(text[pos:]))
    return HTML(f'<div style="white-space:pre-wrap;line-height:1.8">{"".join(out)}</div>')


def compare_f1(results):
    """{alternativa: métricas} -> tabla de F1 lado a lado."""
    names = list(results)
    keys = list(results[names[0]])
    print(f"{'F1':12}" + "".join(f"{n:>16}" for n in names))
    for k in keys:
        print(f"{k:12}" + "".join(f"{results[n][k]['f1']:16.3f}" for n in names))


def show_linked(text, entities):
    display(highlight(text, spans(entities)))
    for e in entities:
        print(f"[{e.label:4}] {e.text[:30]:31} {e.qid:11} {e.wikidata_label} | {e.reasoning}")
```

```python
aida = load_aida("test")
docs = aida[:LIMIT]
with (DATA_DIR / "manual_ner.jsonl").open(encoding="utf-8") as f:
    manual = {
        d["doc_id"]: Document(d["doc_id"], d["text"], [GoldMention(**m) for m in d["mentions"]])
        for d in map(json.loads, f)
    }

# Un solo GLiNER (~2,8 GB) para todo el notebook
pipeline = EntityLinkingPipeline(use_llm=False)
ner = pipeline.ner_service
print(f"AIDA: {len(aida)} documentos (se usan {len(docs)}); manual: {list(manual)}")
```

## 1. NER: GLiNER con tipos finos

**Decisión:** GLiNER zero-shot pidiendo **tipos finos** (`FINE_LABELS`: *sports team*, *country*, *nationality*...) agrupados en PER/ORG/LOC/MISC. El tipo fino se conserva para filtrar candidatos de Wikidata (sección 4).

### 1.1 Tipos finos frente a etiquetas genéricas

```python
# AIDA (noticias): etiquetas genéricas (person, organization...) vs tipos finos
compare_f1({
    "genéricas": evaluate_ner(ner, docs, GENERIC_LABELS)["metrics"],
    "tipos finos": evaluate_ner(ner, docs, FINE_LABELS)["metrics"],
})
```

```python
# Textos anotados a mano: etiquetas adaptadas al dominio de cada texto
CUSTOM_LABELS = {
    "astronomy-1": {
        "person": "PER", "space agency": "ORG", "observatory": "ORG",
        "planet": "LOC", "moon": "LOC", "dwarf planet": "LOC", "star": "LOC",
        "galaxy": "LOC", "nebula": "LOC", "region of the solar system": "LOC",
        "space telescope": "MISC", "space probe": "MISC", "space mission": "MISC",
    },
    "biography-1": {
        "person": "PER", "university": "ORG", "research institute": "ORG",
        "military organization": "ORG", "city": "LOC", "country": "LOC",
        "nationality": "MISC", "award": "MISC", "war": "MISC",
    },
}
for doc_id, d in manual.items():
    print(f"\n== {doc_id}")
    compare_f1({
        "genéricas": evaluate_ner(ner, [d], GENERIC_LABELS)["metrics"],
        "adaptadas": evaluate_ner(ner, [d], CUSTOM_LABELS[doc_id])["metrics"],
    })
```

## 2. Retrieval: `wbsearchentities` con 20 candidatos

**Decisión:** búsqueda por nombre y alias en la API oficial (rápida y estable; SPARQL y la búsqueda semántica se descartaron por latencia, límites e índice desfasado). Se piden 20 candidatos: cuesta lo mismo que 5 y sube el techo del desambiguador (`recall@all`). Se buscan las menciones **anotadas**, para medir solo esta fase.

```python
for limit in (5, 20):
    metrics = (await evaluate_retrieval(docs, limit))["metrics"]
    print(f"{limit:>2} candidatos: " + ", ".join(f"{k} {v:.3f}" for k, v in metrics.items()))
```

## 3. Desambiguación: puntuación probabilística + LLM en casos inciertos

**Decisión:** cada candidato recibe señales en [0, 1] que se combinan linealmente y un softmax da la **probabilidad** de cada uno:

| Señal | Qué mide | Coste |
|---|---|---|
| `prior` | posición en la búsqueda de Wikidata (popularidad) | gratis |
| `context` | palabras de la descripción del candidato cerca de la mención | gratis |
| `type` | compatibilidad con el tipo fino del NER (P31/P279*) | 1 consulta SPARQL por lote |
| `coherence` | clases compartidas con las entidades ya resueltas del documento | idem |
| `llm` | elección del LLM (prompt de elección múltiple) | ~7 s por mención |

**Carga diferida:** las clases solo se piden para las menciones con probabilidad < 0,7, y el LLM solo actúa si siguen inciertas. Los pesos se eligieron en rejilla sobre 13 documentos y se validan en otros 17 (partición por documento).

Muestra: menciones de AIDA donde GLiNER acierta los límites y el QID correcto está entre los candidatos (`scripts/build_disambiguation_sample.py`). `rank 1` = fácil; `rank 2-10` = objetivo. `confident`/`uncertain`: acierto con confianza ≥ / < 0,7.

```python
import zlib

disamb = load_disambiguation_sample()
# Partición por documento: pesos elegidos en train, cifras sobre test
test = [r for r in disamb if zlib.crc32(r["doc_id"].encode()) % 2 == 1]
dis = pipeline.disambiguator_service

print(f"{'1er candidato':18} all {sum(r['gold_rank'] == 1 for r in test) / len(test):.3f}")
for name, use_types in [("señales sin red", False), ("+tipo+coherencia", True)]:
    dis.use_types = use_types
    result = await evaluate_disambiguation(test, dis, use_llm=False)
    print(f"{name:18}", {k: v for k, v in result["accuracy"].items()})
dis.use_types = True
```

```python
# + LLM (qwen2.5:3b) solo en las menciones que siguen inciertas
if RUN_LLM:
    result = await evaluate_disambiguation(test, dis, use_llm=True)
    print(result["accuracy"])
    print(f"{result['llm_calls']} llamadas al LLM de {len(test)} menciones, {result['seconds']:.0f} s")
```

## 4. End-to-end

Pipeline completo (NER + retrieval + desambiguación, sin LLM para que sea barato) frente al gold de AIDA. Métricas estándar de entity linking (GERBIL / TAC-KBP) con `nervaluate` y el QID como etiqueta: `mention` (límites), `all` (límites + QID), `linked` (solo con QID), `nil` (entidades fuera de Wikidata).

```python
results = {}
for name, use_types in [("sin tipo", False), ("tipo+coherencia", True)]:
    dis.use_types = use_types
    results[name] = (await evaluate_linking(docs, pipeline))["metrics"]
dis.use_types = True
compare_f1(results)
```

## 5. Texto libre: pipeline completo

Texto enciclopédico (el terreno de GLiNER), en inglés y en español (Wikidata se consulta en el idioma del texto). Sin gold: se revisa a ojo. Con `RUN_LLM = True` las menciones dudosas pasan por el LLM.

```python
pipeline.disambiguator_service.use_llm = RUN_LLM

en_text = (
    "Ada Lovelace was an English mathematician and writer, chiefly known for her work "
    "on Charles Babbage's proposed mechanical general-purpose computer, the Analytical "
    "Engine. Born in London as the only legitimate child of the poet Lord Byron, she was "
    "educated privately by tutors including Augustus De Morgan of University College "
    "London. The British Computer Society awards the Lovelace Medal in her honour."
)
show_linked(en_text, await pipeline.link(en_text))
```

```python
es_text = (
    "Santiago Ramón y Cajal fue un médico español especializado en histología. Nació en "
    "Petilla de Aragón, en Navarra, y estudió Medicina en la Universidad de Zaragoza. En "
    "1906 recibió el Premio Nobel de Fisiología o Medicina junto al italiano Camillo Golgi. "
    "Dirigió el Instituto Cajal, que hoy forma parte del Consejo Superior de "
    "Investigaciones Científicas."
)
show_linked(es_text, await pipeline.link(es_text, language="es"))
```
